## 1. Criando um DataFrame de Exemplo

Vamos criar um DataFrame com dados de vendas de uma loja:

```python
import pandas as pd
import numpy as np

# Dados de vendas
dados_vendas = {
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset', 
                'Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset'],
    'Vendedor': ['Ana', 'João', 'Ana', 'João', 'Maria', 
                 'Maria', 'Ana', 'João', 'Maria', 'Ana'],
    'Quantidade': [5, 10, 8, 3, 15, 7, 12, 6, 4, 9],
    'Preco': [2500, 80, 200, 1200, 300, 2500, 80, 200, 1200, 300],
    'Data': ['2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04', '2024-01-05',
             '2024-01-06', '2024-01-07', '2024-01-08', '2024-01-09', '2024-01-10']
}

df_vendas = pd.DataFrame(dados_vendas)
df_vendas['Total'] = df_vendas['Quantidade'] * df_vendas['Preco']
df_vendas['Data'] = pd.to_datetime(df_vendas['Data'])

print(df_vendas)
```

## 2. Filtros e Seleção Condicional

Filtrar dados é uma das operações mais comuns:

```python
# Filtrar vendas com quantidade > 8
vendas_altas = df_vendas[df_vendas['Quantidade'] > 8]
print("Vendas com quantidade > 8:")
print(vendas_altas)

# Filtrar vendas do produto 'Notebook'
vendas_notebook = df_vendas[df_vendas['Produto'] == 'Notebook']
print("\nVendas de Notebook:")
print(vendas_notebook)

# Filtro com múltiplas condições
filtro = df_vendas[(df_vendas['Quantidade'] > 5) & (df_vendas['Preco'] > 500)]
print("\nVendas com quantidade > 5 e preço > 500:")
print(filtro)
```

## 3. Ordenação

Ordenar dados por uma ou mais colunas:

```python
# Ordenar por Total (decrescente)
df_ordenado = df_vendas.sort_values('Total', ascending=False)
print("Vendas ordenadas por Total (maior para menor):")
print(df_ordenado)

# Ordenar por múltiplas colunas
df_ordenado2 = df_vendas.sort_values(['Vendedor', 'Total'], ascending=[True, False])
print("\nVendas ordenadas por Vendedor e Total:")
print(df_ordenado2)
```

## 4. GroupBy - Agrupamento de Dados

Agrupar dados e aplicar funções de agregação:

```python
# Total de vendas por produto
vendas_por_produto = df_vendas.groupby('Produto')['Total'].sum()
print("Total de vendas por produto:")
print(vendas_por_produto)

# Média de quantidade por vendedor
media_por_vendedor = df_vendas.groupby('Vendedor')['Quantidade'].mean()
print("\nMédia de quantidade vendida por vendedor:")
print(media_por_vendedor)

# Múltiplas agregações
agregado = df_vendas.groupby('Vendedor').agg({
    'Total': ['sum', 'mean', 'max'],
    'Quantidade': 'sum'
})
print("\nEstatísticas por vendedor:")
print(agregado)
```

## 5. Tratamento de Dados Faltantes

Valores nulos (NaN) são comuns em dados reais:

```python
# Criando DataFrame com valores faltantes
df_nan = pd.DataFrame({
    'Nome': ['Ana', 'João', 'Maria', None, 'Pedro'],
    'Idade': [25, None, 28, 35, None],
    'Nota': [8.5, 9.0, None, 9.5, 8.0]
})

print("DataFrame com valores faltantes:")
print(df_nan)

# Verificando valores nulos
print("\nValores nulos por coluna:")
print(df_nan.isnull().sum())

# Removendo linhas com valores nulos
df_limpo = df_nan.dropna()
print("\nDataFrame sem valores nulos:")
print(df_limpo)

# Preenchendo valores nulos
df_preenchido = df_nan.fillna({'Idade': 30, 'Nota': 7.0, 'Nome': 'Desconhecido'})
print("\nDataFrame com valores preenchidos:")
print(df_preenchido)
```

## 6. Aplicando Funções

Aplique funções personalizadas aos dados:

```python
# Função para classificar o preço
def classificar_preco(preco):
    if preco < 200:
        return 'Baixo'
    elif preco < 800:
        return 'Médio'
    else:
        return 'Alto'

# Aplicando a função
df_vendas['Categoria_Preco'] = df_vendas['Preco'].apply(classificar_preco)
print(df_vendas[['Produto', 'Preco', 'Categoria_Preco']])
```

### Exercício 2: Agrupamento
Com o DataFrame df_vendas, calcule:
1. Total de vendas por vendedor
2. Quantidade média vendida por produto
3. Preço médio de cada produto
4. Número total de transações por vendedor

```python
import pandas as pd

dados = {
    'Vendedor': ['Ana', 'Carlos', 'Ana', 'Beatriz', 'Ana', 'Carlos'],
    'Produto': ['Teclado', 'Mouse', 'Monitor', 'Teclado', 'Mouse', 'Gabinete'],
    'Preco': [150, 250, 1200, 300, 180, 800],
    'Quantidade': [2, 5, 1, 3, 4, 2]
}
df_vendas = pd.DataFrame(dados)

total_vendas = df_vendas.groupby('Vendedor')['Preco'].sum()
qtd_media = df_vendas.groupby('Produto')['Quantidade'].mean()
preco_medio = df_vendas.groupby('Produto')['Preco'].mean()
transacoes = df_vendas.groupby('Vendedor').size()


print("1. Total por vendedor:\n", total_vendas, "\n")
print("2. Quantidade média por produto:\n", qtd_media, "\n")
print("3. Preço médio por produto:\n", preco_medio, "\n")
print("4. Transações por vendedor:\n", transacoes)
```

### Exercício 3: Tratamento de Dados Faltantes
Crie um DataFrame com dados de alunos (Nome, Idade, Nota, Faltas) e:
1. Identifique valores nulos
2. Remova linhas com mais de 2 valores nulos
3. Preencha os valores nulos restantes com a média de cada coluna

```python
import pandas as pd
import numpy as np

dados = {
    'Nome': ['Ana', 'Bia', 'Caio', 'Daniel', 'Eva'],
    'Idade': [15, np.nan, 16, np.nan, 17],
    'Nota': [8.5, 7.0, np.nan, np.nan, 9.0],
    'Faltas': [2, np.nan, 1, np.nan, 0]
}
df_alunos = pd.DataFrame(dados)

nulos = df_alunos.isnull().sum()
df_filtrado = df_alunos[df_alunos.isnull().sum(axis=1) <= 2].copy()
cols_num = df_filtrado.select_dtypes(include=[np.number]).columns
df_filtrado[cols_num] = df_filtrado[cols_num].fillna(df_filtrado[cols_num].mean())

print("--- Quantidade de Nulos por Coluna ---")
print(nulos, "\n")

print("--- DataFrame Tratado ---")
print(df_filtrado)
```

### Exercício 4: Aplicando Funções
Crie uma função que classifique os vendedores em 'Estrela' (total > 5000), 'Bom' (total > 2000) e 'Regular' (total <= 2000).

Aplique esta função ao DataFrame df_vendas e adicione uma nova coluna com a classificação.

```python
import pandas as pd

dados = {
    'Vendedor': ['Ana', 'Carlos', 'Beatriz', 'Daniel'],
    'Total_Vendas': [6500, 3200, 1500, 2000]
}
df_vendas = pd.DataFrame(dados)
def classificar(total):
    if total > 5000:
        return 'Estrela'
    elif total > 2000:
        return 'Bom'
    else:
        return 'Regular'
df_vendas['Classificacao'] = df_vendas['Total_Vendas'].apply(classificar)

print(df_vendas)
```
